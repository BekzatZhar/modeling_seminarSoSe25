```python
import numpy as np
import matplotlib.pyplot as plt
import gudhi as gd
from gudhi.hera import wasserstein_distance
from ripser import ripser
```

```python
points1=np.load('0b_pp.npy')
result1 = ripser(points1)
diagrams1 = result1['dgms']

points2=np.load('0_pp.npy')
result2 = ripser(points2)
diagrams2 = result2['dgms']

dist_h1 = wasserstein_distance(diagrams1[1], diagrams2[1], order=1)
dist_h0 = wasserstein_distance(diagrams1[0], diagrams2[0], order=1)
print(dist_h1, dist_h0)
```

```python
n = 10

wasserstein_matrix_b_h1 = np.zeros((n, n))
wasserstein_matrix_h1 = np.zeros((n, n))

wasserstein_matrix_b_h0 = np.zeros((n, n))
wasserstein_matrix_h0 = np.zeros((n, n))

pps = []
pps_b = []
for i in range(n):
    points = np.load(f'{i}_pp.npy')
    pps.append(points)
    points_b = np.load(f'{i}b_pp.npy')
    pps_b.append(points_b)

for i in range(n):
    for j in range(n):

        result_1 = ripser(pps[i])['dgms']
        
        result_2 = ripser(pps[j])['dgms']
        
        result_b = ripser(pps_b[j])['dgms']

        distance = wasserstein_distance(result_1[1], result_b[1], order=1)
        wasserstein_matrix_b_h1[i, j] = distance
        
        distance = wasserstein_distance(result_1[1], result_2[1], order=1)
        wasserstein_matrix_h1[i, j] = distance
        
        distance = wasserstein_distance(result_1[0], result_b[0], order=1)
        wasserstein_matrix_b_h0[i, j] = distance
        
        distance = wasserstein_distance(result_1[0], result_2[0], order=1)
        wasserstein_matrix_h0[i, j] = distance
```

```python
print("---== mean and var of pairwise no bump ==---")
print("h_0: ")
print(np.mean(wasserstein_matrix_h0))
print(np.var(wasserstein_matrix_h0))
print("h_1: ")
print(np.mean(wasserstein_matrix_h1))
print(np.var(wasserstein_matrix_h1))

print("---== mean and var of bump vs no bump ==---")
print("h_0: ")
print(np.mean(wasserstein_matrix_b_h0))
print(np.var(wasserstein_matrix_b_h0))
print("h_1: ")
print(np.mean(wasserstein_matrix_b_h1))
print(np.var(wasserstein_matrix_b_h1))

```
