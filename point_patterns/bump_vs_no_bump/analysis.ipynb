```python
import os
import numpy as np
import matplotlib.pyplot as plt
import gudhi as gd
from gudhi.hera import wasserstein_distance
```

```python
n = 10

wasserstein_matrix_b_1 = np.zeros((n, n))
wasserstein_matrix_1 = np.zeros((n, n))

for i in range(n):
    for j in range(n):

        acb = gd.AlphaComplex(points=np.load(f'point_patterns/bump_vs_no_bump/{i}b_pp.npy'))
        stb = acb.create_simplex_tree()
        stb.compute_persistence()
        dgmb = stb.persistence_intervals_in_dimension(1)

        ac1 = gd.AlphaComplex(points=np.load(f'point_patterns/bump_vs_no_bump/{j}_pp.npy'))
        st1 = ac1.create_simplex_tree()
        st1.compute_persistence()
        dgm1 = st1.persistence_intervals_in_dimension(1)
        
        ac2 = gd.AlphaComplex(points=np.load(f'point_patterns/bump_vs_no_bump/{i}_pp.npy'))
        st2 = ac2.create_simplex_tree()
        st2.compute_persistence()
        dgm2 = st2.persistence_intervals_in_dimension(1)

        distance = wasserstein_distance(dgm1, dgmb, order=1)
        wasserstein_matrix_b_1[i, j] = distance
        
        distance = wasserstein_distance(dgm1, dgm2, order=1)
        wasserstein_matrix_1[i, j] = distance
```

```python
print(np.mean(wasserstein_matrix_1))
print(np.var(wasserstein_matrix_1))

print(np.mean(wasserstein_matrix_b_1))
print(np.var(wasserstein_matrix_b_1))

```
